# E1. Diagnóstico y calidad de datos

### Objetivos:
- **Perfilamiento**: volumen, completitud, distribución y consistencia de cada fuente, y entre sí.
- **Hallazgos**: problemas con evidencia cuantitativa y su impacto.
- **Reglas de calidad**: reglas para que el problema no pase desapercibido.
- **Preguntas al dueño del dato**: máximo 8, priorizadas.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

# Configuraciones visuales
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
```

```python
# Cargar los datos de las tres IPS
df_norte = pd.read_csv('../dataset/ips_norte_citas.csv')
df_sur = pd.read_csv('../dataset/ips_sur_citas.csv')

# Hallazgo E1: Occidente usa punto y coma como separador.
df_occidente = pd.read_csv('../dataset/ips_occidente_citas.csv', sep=';')

# Cargar los eventos de WhatsApp
eventos_wpp = []
with open('../dataset/whatsapp_eventos.jsonl', 'r') as f:
    for line in f:
        eventos_wpp.append(json.loads(line))

df_wpp = pd.DataFrame(eventos_wpp)

print(f"Volumen IPS Norte: {df_norte.shape}")
print(f"Volumen IPS Sur: {df_sur.shape}")
print(f"Volumen IPS Occidente: {df_occidente.shape}")
print(f"Volumen WhatsApp Eventos: {df_wpp.shape}")
```

```python
# Hallazgo E1: Inconsistencias de esquema y Datos Personales Filtrados
print("\n--- Columnas Norte ---")
print(df_norte.columns.tolist())
print("\n--- Columnas Sur ---")
print(df_sur.columns.tolist())
print("\n--- Columnas Occidente ---")
print(df_occidente.columns.tolist())

print("\n\n¡Alerta! Columnas extra en Sur:", set(df_sur.columns) - set(df_norte.columns))
```
